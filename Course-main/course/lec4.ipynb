```python
import cv2

# Load the video
video_path = r"\\192.168.1.10\shear all\day4\british_highway_traffic.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
else:
    # Video information
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = frame_count / fps if fps > 0 else 0

    print("Video opened successfully!")
    print("-----------------------------")
    print(f"Width: {width} pixels")
    print(f"Height: {height} pixels")
    print(f"FPS: {fps:.2f}")
    print(f"Frame count: {frame_count}")
    print(f"Duration: {duration:.2f} seconds")
    print("-----------------------------")

    frame_number = 0

    while True:
        ret, frame = cap.read()

        if not ret:
            break

        # ------------------------------------------------
        # 1. Grayscale conversion
        # ------------------------------------------------
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

        # ------------------------------------------------
        # 2. Gaussian filtering
        # ------------------------------------------------
        gaussian = cv2.GaussianBlur(gray, (5, 5), 0)

        # Median filtering
        median = cv2.medianBlur(gray, 5)

        # ------------------------------------------------
        # 3. Canny edge detection
        # ------------------------------------------------
        canny = cv2.Canny(gaussian, 50, 150)

        # ------------------------------------------------
        # 4. Sobel edge detection
        # ------------------------------------------------
        sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
        sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

        sobel_x = cv2.convertScaleAbs(sobel_x)
        sobel_y = cv2.convertScaleAbs(sobel_y)

        sobel = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

        # ------------------------------------------------
        # 5. Scharr edge detection
        # ------------------------------------------------
        scharr_x = cv2.Scharr(gray, cv2.CV_64F, 1, 0)
        scharr_y = cv2.Scharr(gray, cv2.CV_64F, 0, 1)

        scharr_x = cv2.convertScaleAbs(scharr_x)
        scharr_y = cv2.convertScaleAbs(scharr_y)

        scharr = cv2.addWeighted(scharr_x, 0.5, scharr_y, 0.5, 0)

        # ------------------------------------------------
        # 6. Global thresholding
        # ------------------------------------------------
        _, global_thresh = cv2.threshold(
            gray, 127, 255, cv2.THRESH_BINARY
        )

        # ------------------------------------------------
        # 7. Otsu thresholding
        # ------------------------------------------------
        _, otsu = cv2.threshold(
            gray, 0, 255,
            cv2.THRESH_BINARY + cv2.THRESH_OTSU
        )

        # ------------------------------------------------
        # 8. Adaptive thresholding
        # ------------------------------------------------
        adaptive = cv2.adaptiveThreshold(
            gray,
            255,
            cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
            cv2.THRESH_BINARY,
            11,
            2
        )

        # ------------------------------------------------
        # 9. Morphological operations
        # ------------------------------------------------
        kernel = cv2.getStructuringElement(
            cv2.MORPH_RECT, (5, 5)
        )

        # Erosion
        erosion = cv2.erode(gray, kernel, iterations=1)

        # Dilation
        dilation = cv2.dilate(gray, kernel, iterations=1)

        # Opening
        opening = cv2.morphologyEx(
            gray, cv2.MORPH_OPEN, kernel
        )

        # Closing
        closing = cv2.morphologyEx(
            gray, cv2.MORPH_CLOSE, kernel
        )

        # ------------------------------------------------
        # Display
        # ------------------------------------------------

        cv2.imshow("Original", frame)
        cv2.imshow("Grayscale", gray)
        cv2.imshow("Gaussian Blur", gaussian)
        cv2.imshow("Median Blur", median)
        cv2.imshow("Canny", canny)
        cv2.imshow("Sobel", sobel)
        cv2.imshow("Scharr", scharr)
        cv2.imshow("Global Threshold", global_thresh)
        cv2.imshow("Otsu", otsu)
        cv2.imshow("Adaptive Threshold", adaptive)
        cv2.imshow("Erosion", erosion)
        cv2.imshow("Dilation", dilation)
        cv2.imshow("Opening", opening)
        cv2.imshow("Closing", closing)

        # Print current frame
        print(f"Current frame: {frame_number}")
        frame_number += 1

        # ESC or q to stop
        key = cv2.waitKey(30) & 0xFF

        if key == 27 or key == ord('q'):
            break

# Release video
cap.release()
cv2.destroyAllWindows()


```

```python
import cv2

# قراءة الصورة
image = cv2.imread(r"FRAMES\frame_01412.jpg")

# 1. تحويل إلى Grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# 2. تقدير الخلفية وطرحها
background = cv2.GaussianBlur(gray, (51, 51), 0)
foreground = cv2.absdiff(gray, background)

# 3. Threshold
_, binary = cv2.threshold(
    foreground,
    30,
    255,
    cv2.THRESH_BINARY
)

# تنظيف الصورة من الضوضاء
kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))
binary = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
binary = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

# 4. إيجاد الـ Contours
contours, _ = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# 5. حذف الأجسام الصغيرة
min_area = 500
count = 0

for contour in contours:

    if cv2.contourArea(contour) < min_area:
        continue

    # 6. رسم Bounding Rectangle
    x, y, w, h = cv2.boundingRect(contour)

    cv2.rectangle(
        image,
        (x, y),
        (x + w, y + h),
        (0, 255, 0),
        2
    )

    count += 1

# 7. عرض عدد الأجسام
cv2.putText(
    image,
    f"Detected objects: {count}",
    (20, 40),
    cv2.FONT_HERSHEY_SIMPLEX,
    1,
    (0, 0, 255),
    2
)

# عرض النتائج
cv2.imshow("Foreground", foreground)
cv2.imshow("Binary", binary)
cv2.imshow("Detected Objects", image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2
import math

# ------------------------------------------------
# Load the video
# ------------------------------------------------
video_path = r"C:\Users\bofa2\OneDrive\Desktop\days\british_highway_traffic.mp4"
cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error: Could not open the video.")
    exit()

# ------------------------------------------------
# Video information
# ------------------------------------------------
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

duration = frame_count / fps if fps > 0 else 0

print("Video opened successfully!")
print("-----------------------------")
print(f"Width: {width} pixels")
print(f"Height: {height} pixels")
print(f"FPS: {fps:.2f}")
print(f"Frame count: {frame_count}")
print(f"Duration: {duration:.2f} seconds")
print("-----------------------------")


# ------------------------------------------------
# Background Subtraction
# ------------------------------------------------
background_subtractor = cv2.createBackgroundSubtractorMOG2(
    history=500,
    varThreshold=50,
    detectShadows=True
)


# ------------------------------------------------
# Object tracking variables
# ------------------------------------------------
next_object_id = 0

# Dictionary:
# ID -> (center_x, center_y)
objects = {}

# Maximum distance allowed between old and new center
max_distance = 50

# Minimum object area
min_area = 500


# ------------------------------------------------
# Function to calculate distance
# ------------------------------------------------
def calculate_distance(point1, point2):
    return math.sqrt(
        (point1[0] - point2[0]) ** 2 +
        (point1[1] - point2[1]) ** 2
    )


# ------------------------------------------------
# Read video continuously
# ------------------------------------------------
frame_number = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # ------------------------------------------------
    # 1. Background Subtraction
    # ------------------------------------------------
    fg_mask = background_subtractor.apply(frame)

    # ------------------------------------------------
    # 2. Remove shadows and noise
    # ------------------------------------------------
    _, fg_mask = cv2.threshold(
        fg_mask,
        200,
        255,
        cv2.THRESH_BINARY
    )

    # ------------------------------------------------
    # 3. Morphological operations
    # ------------------------------------------------
    kernel = cv2.getStructuringElement(
        cv2.MORPH_ELLIPSE,
        (5, 5)
    )

    # Remove small noise
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_OPEN,
        kernel
    )

    # Connect nearby regions
    fg_mask = cv2.morphologyEx(
        fg_mask,
        cv2.MORPH_CLOSE,
        kernel
    )

    # Expand objects slightly
    fg_mask = cv2.dilate(
        fg_mask,
        kernel,
        iterations=2
    )


    # ------------------------------------------------
    # 4. Find contours
    # ------------------------------------------------
    contours, _ = cv2.findContours(
        fg_mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )


    # ------------------------------------------------
    # Store detected objects in current frame
    # ------------------------------------------------
    current_objects = []

    for contour in contours:

        # Calculate area
        area = cv2.contourArea(contour)

        # Ignore small objects/noise
        if area < min_area:
            continue

        # Bounding box
        x, y, w, h = cv2.boundingRect(contour)

        # Center of object
        center_x = x + w // 2
        center_y = y + h // 2

        current_objects.append(
            (center_x, center_y, x, y, w, h)
        )


    # ------------------------------------------------
    # 5. Object Tracking
    # ------------------------------------------------

    new_objects = {}

    # Keep track of IDs that have already been used
    used_ids = set()

    for center_x, center_y, x, y, w, h in current_objects:

        current_center = (center_x, center_y)

        best_id = None
        best_distance = max_distance

        # Compare with previous objects
        for object_id, previous_center in objects.items():

            if object_id in used_ids:
                continue

            distance = calculate_distance(
                current_center,
                previous_center
            )

            if distance < best_distance:
                best_distance = distance
                best_id = object_id

        # If no matching object was found,
        # create a new ID
        if best_id is None:

            best_id = next_object_id
            next_object_id += 1

        # Save object
        new_objects[best_id] = current_center
        used_ids.add(best_id)

        # ------------------------------------------------
        # 6. Draw Bounding Box
        # ------------------------------------------------
        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        # Draw center point
        cv2.circle(
            frame,
            (center_x, center_y),
            5,
            (0, 0, 255),
            -1
        )

        # Display object ID
        cv2.putText(
            frame,
            f"ID: {best_id}",
            (x, y - 10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )


    # Update objects
    objects = new_objects


    # ------------------------------------------------
    # 7. Display frame number
    # ------------------------------------------------
    cv2.putText(
        frame,
        f"Frame: {frame_number}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # Number of detected objects
    cv2.putText(
        frame,
        f"Objects: {len(current_objects)}",
        (20, 80),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (255, 255, 0),
        2
    )


    # ------------------------------------------------
    # 8. Display results
    # ------------------------------------------------

    cv2.imshow("Object Detection and Tracking", frame)

    cv2.imshow("Background Subtraction", fg_mask)


    print(
        f"Frame: {frame_number} | "
        f"Objects detected: {len(current_objects)}"
    )

    frame_number += 1


    # ------------------------------------------------
    # 9. ESC or q to stop
    # ------------------------------------------------
    key = cv2.waitKey(30) & 0xFF

    if key == 27 or key == ord('q'):
        break


# ------------------------------------------------
# Release everything
# ------------------------------------------------
cap.release()
cv2.destroyAllWindows()

print("Finished.")

```
